```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
X=df.drop('Y',axis=1)
y=df['Y']
```

```python
le=LabelEncoder()
y=le.fit_transform(y)
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.25,random_state=42)

print("train size:", X_train.shape)
print("test size:", X_test.shape)
```

# Model Implementation

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([
    Dense(64, activation='relu', input_shape=(60,)),
    Dense(32, activation='relu'),
    Dense(1,  activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
history=model.fit(X_train,y_train,epochs=100,batch_size=32,validation_split=0.2, verbose=0)
```

```python
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')
plt.title('Model Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
y_pred=(model.predict(X_test)>0.5).astype(int)
loss,acc=model.evaluate(X_test,y_test,verbose=0)
print("Default Model Accuracy:",round(acc, 4))
```

# Hyperparameter Tuning

```python
import itertools
from tensorflow.keras.optimizers import Adam
```

```python
layers_options=[[64], [128, 64], [128, 64, 32]]
activation_options=['relu', 'tanh']
lr_options=[0.01, 0.001, 0.0001]
```

```python
results = []
combos = list(itertools.product(layers_options, activation_options, lr_options))
print(f"Testing {len(combos)} combinations...")
```

```python
for layers, act, lr in combos:
    m = Sequential()
    m.add(Dense(layers[0], activation=act, input_shape=(60,)))
    for n in layers[1:]:
        m.add(Dense(n, activation=act))
    m.add(Dense(1, activation='sigmoid'))
    m.compile(optimizer=Adam(learning_rate=lr), loss='binary_crossentropy', metrics=['accuracy'])
    m.fit(X_train, y_train, epochs=50, batch_size=32, verbose=0)
    _, acc = m.evaluate(X_test, y_test, verbose=0)
    results.append({'layers': layers, 'activation': act, 'lr': lr, 'accuracy': round(acc, 4)})

print("Done.")
```

```python
results_df=pd.DataFrame(results).sort_values('accuracy',ascending=False).reset_index(drop=True)
results_df.head(10)
```

```python
best=results_df.iloc[0]
print("Best layers:", best['layers'])
print("Best activation:", best['activation'])
print("Best lr:", best['lr'])
print("Best accuracy:", best['accuracy'])
```

```python
final=Sequential()
final.add(Dense(best['layers'][0], activation=best['activation'], input_shape=(60,)))
for n in best['layers'][1:]:
    final.add(Dense(n, activation=best['activation']))
final.add(Dense(1, activation='sigmoid'))
final.compile(optimizer=Adam(learning_rate=best['lr']), loss='binary_crossentropy', metrics=['accuracy'])
final.fit(X_train, y_train, epochs=100, batch_size=32, verbose=0)

_,tuned_acc=final.evaluate(X_test, y_test, verbose=0)
print("Tuned Model Accuracy:", round(tuned_acc, 4))
```

# Evaluation

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred_tuned=(final.predict(X_test) > 0.5).astype(int)
print(classification_report(y_test, y_pred_tuned, target_names=le.classes_))
```

```python
cm = confusion_matrix(y_test, y_pred_tuned)
sns.heatmap(cm, annot=True, fmt='d', xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix — Tuned Model')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
